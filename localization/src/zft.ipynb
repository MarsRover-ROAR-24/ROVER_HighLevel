```python
#!/usr/bin/env python3
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
path = 2
cap = cv2.VideoCapture(path)
while True: 
    ret, frame = cap.read()
    frame = cv2.resize(frame, (640, 480))
    if not ret:
        break
    cv2.imshow('frame', frame)
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break
cap.release()
cv2.destroyAllWindows()
```
